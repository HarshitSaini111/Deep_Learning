```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
words = ["BERT", "helps", "sentiment", "analysis"]

words
```

```python
attention = np.array([
    [0.15, 0.35, 0.30, 0.20],
    [0.20, 0.40, 0.20, 0.20],
    [0.10, 0.25, 0.50, 0.15],
    [0.15, 0.20, 0.35, 0.30]
])

attention
```

```python
attention.sum(axis=1)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    attention,
    xticklabels=words,
    yticklabels=words,
    annot=True,
    fmt='.2f',
    cmap='Reds'
)

plt.title('BERT Attention Heatmap')

plt.xlabel('Key (Words being attended to)')

plt.ylabel('Query (Current word)')

plt.tight_layout()

plt.show()
```

```python
for i, query_word in enumerate(words):

    max_index = np.argmax(attention[i])

    print(
        f'Query word "{query_word}" attends most to "{words[max_index]}" with score {attention[i][max_index]:.2f}'
    )
```
